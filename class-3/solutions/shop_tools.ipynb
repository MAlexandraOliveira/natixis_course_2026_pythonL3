# 🧩 `shop_tools` — solution to exercise D1

The layer on top. It imports `products` and adds the shop's own rules — the VAT rate, what
counts as low stock, and how a report line looks. Nothing here recalculates what `products`
already knows.

```python
"""Shop rules, built on top of the products module."""

import import_ipynb
import products

VAT_RATE = 23
LOW_STOCK = 10


def add_vat(amount, rate=VAT_RATE):
    """Return the amount with VAT added, rounded to 2 decimals."""
    return round(amount + amount * rate / 100, 2)


def is_low(item):
    """True when there are fewer than LOW_STOCK units in stock."""
    return item.quantity < LOW_STOCK


def report_line(item):
    """One line of the stock report, marked when stock is low."""
    line = f"{item.name:<10}{item.quantity:>4}{products.format_eur(item.value()):>15}"
    if is_low(item):
        line = line + "   LOW"
    return line


def stock_summary(items):
    """Everything the report needs: what products knows, plus VAT and the low count."""
    summary = products.summarise(items)
    summary["with_vat"] = add_vat(summary["value"])
    summary["low"] = len([item for item in items if is_low(item)])
    return summary
```
