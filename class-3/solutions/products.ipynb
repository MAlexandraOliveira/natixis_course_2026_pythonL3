# 🧩 `products` — solution to exercise C1

A module that imports `statistics` for itself, holds a class, and lets its own functions call
each other.

```python
"""Products, and what they are worth."""

import statistics


def format_eur(amount):
    """Return the amount as a tidy string with thousands separators."""
    return f"{amount:,.2f} EUR"


class Product:
    """One product line: a name, a price and how many are in stock."""

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def value(self):
        """Price times quantity, rounded to 2 decimals."""
        return round(self.price * self.quantity, 2)

    def describe(self):
        """One line of a report - format_eur() is in this same module."""
        return f"{self.name:<10}{self.quantity:>4} x {format_eur(self.price):>12}"


def summarise(items):
    """Return a dict of how many products, their total value and the average price."""
    values = [item.value() for item in items]
    prices = [item.price for item in items]
    return {
        "products": len(items),
        "value": round(sum(values), 2),
        "average_price": round(statistics.mean(prices), 2),
    }
```
