# 🧩 `payment_tools` — the toolkit for section 7

Everything the payments report needs, in one place: a class, two functions and a value. The
class notebook imports it three different ways and builds a report **without repeating a single
rule**.

Keep this notebook in the same folder as the class notebook.

> 💡 Notice what is *not* here: no `print`, no report, no data. A module holds the rules and the
> tools. The notebook that imports it holds the data and decides what to show.

```python
"""Everything needed to work with payments: a Payment class and two helpers."""

import statistics

VAT_RATE = 23


class Payment:
    """One payment to a supplier."""

    def __init__(self, reference, supplier, amount):
        self.reference = reference
        self.supplier = supplier
        self.amount = amount

    def total(self):
        """What this payment is worth, in EUR."""
        return round(self.amount, 2)

    def describe(self):
        """One line of a report."""
        return f"{self.reference}  {self.supplier:<16}{format_eur(self.total()):>14}"


def format_eur(amount):
    """Return the amount as a tidy string."""
    return f"{amount:,.2f} EUR"


def summarise(payments):
    """Return a dict of the count, total and average of some payments."""
    totals = [payment.total() for payment in payments]
    return {
        "count": len(totals),
        "total": round(sum(totals), 2),
        "average": round(statistics.mean(totals), 2),
    }
```
