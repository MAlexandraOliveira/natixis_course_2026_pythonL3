# 🧩 `money_tools` — a module, written as a notebook

This notebook does not *do* anything. It **defines** things — a value and two functions — for
another notebook to borrow.

- The class notebook imports it with `import money_tools` (see section 4 of Level 3 Class 3).
- Keep this file **in the same folder** as the notebook that imports it.
- Run it on its own and the self-test at the bottom runs. Import it and that block is skipped.

> 💡 A module is just reusable code you write once and use somewhere else. There is nothing
> special about this notebook — it is only the *place* the code lives.

```python
"""Small helpers for working with money."""

VAT_RATE = 23


def add_vat(amount, rate=VAT_RATE):
    """Return the amount with VAT added, rounded to 2 decimals."""
    return round(amount + amount * rate / 100, 2)


def format_eur(amount):
    """Return the amount as a tidy string with thousands separators."""
    return f"{amount:,.2f} EUR"
```

### The self-test

The block below runs **only when you run this notebook directly**. When another notebook
imports `money_tools`, it is skipped — so importing stays silent, which is what you want from a
module.

`__name__` is a name Python sets for you: it is `"__main__"` in the notebook you actually ran,
and the module's own name everywhere else.

```python
if __name__ == "__main__":
    print("Running money_tools directly - self-test:")
    print(add_vat(100))          # 123.0
    print(add_vat(100, rate=6))  # 106.0
    print(format_eur(1234.5))    # 1,234.50 EUR
```
